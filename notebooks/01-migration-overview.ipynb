# Mapping migration

### An introduction to vector data operations

The Veery thrush (*Catharus fuscescens*) migrates each year between nesting sites in North America and South America. Veeries are small birds, and encountering a hurricane during their migration would be disasterous. However, Ornithologist Christopher Heckscher has found in tracking their migration that their breeding patterns and migration timing can predicting the severity of hurricane season as accurately as meterological models [(Heckscher 2018)](https://doi.org/10.1038/s41598-018-28302-3)!

**Your answer here:**

Animals migrate for access to food, better seasonal conditions, and for reproduction. 

Climate change is leading to more frequent and severe storms. If animals have to migrate through these storms, their lives could be put at risk. Additionally, they may have to adjust their migration patterns to avoid these severe storms. The food they rely on may also become more difficult to access or more sparse. 

I used to notice geese in Denver during warmer months, but they would be gone for the winter. Now, they seem to be in Denver almost year around. The Denverite reports that this is because we have created parks with plenty of grass for them to feed on that they do not need to travel such far distances for their food anymore. While this is not directly connected to climate change, this example shows how human decisions and behaviors can impact animal migration.
